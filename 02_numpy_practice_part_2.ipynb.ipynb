```python
import random # import the Random Library 
def generated_otp(lenght=4): # define Function(generated otp) & lenght
    """Generated a numeric OTP of specified lenght""" #'''This is called a docstring.It is used to describe what the function does
    digits = '012345'
    otp = ''.join( random.choice(digits)for _ in range (lenght))# This picks one random character from dgits & This repeats the random selection.
    return otp # Join function used to join  otherwise they show '2','4','1','5'
otp_lenght = 4 # you can change this to any lenght you prefere
otp = generated_otp(otp_lenght)
print(f"your otp is : {otp}")# you can change this to any lenght you prefere
```
